### Import bibliotek 

```python
import os
import json
import string
import morfeusz2
import numpy as np
import pandas as pd
import operator as op
import itertools as it
from wordcloud import WordCloud
from matplotlib import pyplot as plt
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
from nltk.corpus import PlaintextCorpusReader
from scipy.cluster.hierarchy import dendrogram
from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import LatentDirichletAllocation, NMF
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

1. Utworzenie korpusu dokumentów

```python
corpus_dir="./literatura"
corpus=PlaintextCorpusReader(corpus_dir, ".*\.txt")
files= corpus.fileids()
files

```

2. Wstępne przygotowanie dokumentów

```python
documents={}
for f in files:
    documents[f]=corpus.raw(f)

print(json.dumps(documents, indent=4, ensure_ascii=False))
```

```python
stoplist_file=open("./stopwords_pl.txt", "r", encoding="UTF-8")
stoplist=stoplist_file.read().splitlines()
stoplist_file.close()
stoplist=stoplist[4:]
stoplist
```

```python
def lemmatize(text):
    morf=morfeusz2.Morfeusz()
    segments=it.groupby(morf.analyse(text), op.itemgetter(0,1))
    def disambiguate(group):
        pairs=((len(text), lemma)
               for _, _, (_, lemma, text, _, _,) in group)
        perpl, lemma=min(pairs)
        return lemma.split(":")[0]
    lemmas=(disambiguate(group) for key, group in segments)
    return " ".join(filter(str.isalpha, lemmas))
```

```python
for key in documents:
    documents[key]=documents[key].lower()
    documents[key]="".join([char for char in documents[key] if char not in string.punctuation])
    documents[key]=lemmatize(documents[key])
    documents[key]=" ".join([word for word in word_tokenize(documents[key], language='polish') if word not in stoplist])
print(json.dumps(documents, indent=4, ensure_ascii=False))
```

3. Utworzenie macierzy częstości

```python
docs=pd.DataFrame.from_dict(documents, orient='index', columns=['text'])
docs.columns=['content']
docs
```

```python


count_vectorizer=CountVectorizer()
counts_tf= count_vectorizer.fit_transform(docs['content'])
counts_tf.toarray()
```

```python
tfidf_vectorizer=TfidfVectorizer()
counts_tfidf= tfidf_vectorizer.fit_transform(docs['content'])
counts_tfidf.toarray()
```

5. Topic modelling

```python
def plot_top_words(model, feature_names, n_top_words, title, subplots):
    fig, axes = plt.subplots(*subplots, figsize=(30, 15), sharex=True)
    axes = axes.flatten()
    for topic_idx, topic in enumerate(model.components_):
        top_features_ind = topic.argsort()[-n_top_words:]
        top_features = feature_names[top_features_ind]
        weights = topic[top_features_ind]

        ax = axes[topic_idx]
        ax.barh(top_features, weights, height=0.7)
        ax.set_title(f"Topic {topic_idx +1}", fontdict={"fontsize": 30})
        ax.tick_params(axis="both", which="major", labelsize=20)
        for i in "top right left".split():
            ax.spines[i].set_visible(False)
        fig.suptitle(title, fontsize=40)

    plt.subplots_adjust(top=0.90, bottom=0.05, wspace=0.90, hspace=0.3)
    if not os.path.exists("./tematy"):
        os.makedirs("./tematy")
    plt.savefig(f"./tematy/{title}.png")
    plt.show()

    
```

```python
n_components=6
feature_names = count_vectorizer.get_feature_names_out()
n_top_words = 20
subplots=(2, 3)
```

```python
from sklearn.decomposition import LatentDirichletAllocation

lda = LatentDirichletAllocation(
    n_components=n_components,
    max_iter=5,
    learning_method="online",
    learning_offset=50.0,
    random_state=0
)

lda.fit(counts_tf)
plot_top_words(
    lda,
    feature_names,
    n_top_words,
    "Tematy w modelu LDA",
    subplots
)
```

```python
nmf_fn= NMF(
    n_components=n_components,
    random_state=1,
    alpha_H=0.00005,
    alpha_W=.000005,
    l1_ratio=.5,
)

nmf_fn.fit(counts_tfidf)
plot_top_words(
    nmf_fn,
    feature_names,
    n_top_words,
    "Tematy w modelu NMF (FN)",
    subplots
)
```

```python
nmf_k1= NMF(
    n_components=n_components,
    random_state=1,
    beta_loss="kullback-leibler",
    solver="mu",
    max_iter=1000,
    alpha_H=0.00005,
    alpha_W=.000005,
    l1_ratio=.5,
)

nmf_k1.fit(counts_tfidf)
plot_top_words(
    nmf_k1,
    feature_names,
    n_top_words,
    "Tematy w modelu NMF (KL)",
    subplots
)
```

6. Analiza skupień

```python
if not os.path.exists("./skupienia"):
    os.mkdir("./skupienia")
```

```python
n_clusters=3
```

```python
def plot_dendrogram(model, **kwargs):
    counts=np.zeros(model.children_.shape[0])
    n_samples=len(model.labels_)
    for i, merge in enumerate(model.children_):
        current_count=0
        for child_idx in merge:
            if child_idx < n_samples:
                current_count += 1
            else:
                current_count += counts[child_idx - n_samples]
        counts[i] = current_count
    linkage_matrix = np.column_stack([model.children_, model.distances_, counts]).astype(float)
    plt.figure(figsize=(7,4))
    dendrogram(linkage_matrix, **kwargs)
```

```python
cs=cosine_similarity(counts_tfidf, counts_tfidf).flatten().reshape(len(files), len(files))
cs
```

```python
clustering =AgglomerativeClustering(
    n_clusters=n_clusters,
    metric='euclidean',
    linkage='ward',
    compute_distances=True
).fit(cs)
clustering.labels_
```

```python
plot_dendrogram(
    clustering,
    labels=files,
    orientation ='right',
)
plt.gcf().subplots_adjust(left=0.65)
plt.savefig(f"./skupienia/cosine_ward.png")
plt.close()
```

```python
ed=euclidean_distances(counts_tf, counts_tf).flatten().reshape(len(files), len(files))
ed
```

```python
clustering=AgglomerativeClustering(
    n_clusters=n_clusters,
    metric='precomputed',
    linkage='complete',
    compute_distances=True
).fit(ed)
clustering.labels_
```

```python
plot_dendrogram(
    clustering,
    labels=files,
    orientation ='right',
)
plt.gcf().subplots_adjust(left=0.65)
plt.savefig(f"./skupienia/euclidean_complete.png")
plt.close()
```
